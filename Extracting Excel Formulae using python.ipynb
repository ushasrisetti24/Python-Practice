# Extracting Excel Cell Formulae using Python

Reference: https://medium.com/programmed-iq/extract-all-excel-formulas-into-json-using-python-a-step-by-step-tutorial-42b73e371d3d

We always knew only to extract values, but very few people know that there is also a way to extract formulaes also from the workbook.

1) Use OpenpyXL to load spreadsheets
2) assign *data_only = False* while loading- this makes sure to that along with cell values, formulae can also taken.

### Code flow
1. Install openpyxl
2. Load workbook
3. Iterate through workbook
4. Detect formula cells
5. Extract formulae
6. Add row and column coordinates as well
7. Save it in JSON/CSV format


```python
"""JSON formatted code=> output as JSON"""

from openpyxl import load_workbook
import pandas as pd
import re
import json 

fileurl= "C:/Users/hp/Downloads/formula-extraction-practice.xlsx"
wb = load_workbook(fileurl, data_only= False)
result={}
for sheet in wb.sheetnames:
    result[sheet]={}
    for row in wb[sheet].iter_rows():
        for cell in row:
            if (isinstance(cell.value,str) and re.match(r"^=", cell.value)):
                result[sheet][cell.coordinate]= cell.value

with open("C:/Users/hp/Downloads/formula.json", "w") as f:
    json.dump(result,f,indent=2)

    
```

```python
"""Pandas formatted code=> output as CSV"""

from openpyxl import load_workbook
import pandas as pd
import re
import csv

fileurl= "C:/Users/hp/Downloads/formula-extraction-practice.xlsx"
wb= load_workbook(fileurl,data_only=False)
res=[]

for sheet in wb.sheetnames:
    for row in wb[sheet].iter_rows():
        for cell in row:
            if (isinstance(cell.value, str) and re.match(r"^=",cell.value)):
                res.append({
                    "sheet": sheet,
                    "Cell coordinates" : cell.coordinate,
                    "Cell formula": "'" + cell.value + "'"
                })
df= pd.DataFrame(res)

df.to_csv("C:/Users/hp/Downloads/formula.csv", index= False)
        
```
