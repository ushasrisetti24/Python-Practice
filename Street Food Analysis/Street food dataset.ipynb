### Highest Priced Entry for each food

```python
Highest_per_food = (
    df.loc[
        df.groupby("Dish Name")["Typical Price (USD)"].idxmax()
    ]
    .sort_values("Typical Price (USD)", ascending=False)
    .reset_index(drop=True)
)
display(Highest_per_food)
```

### nth Highest and nth Lowest Price of each Dish

```python
nth_highest = (
    df[df.groupby("Dish Name")["Typical Price (USD)"]
       .rank(method="dense", ascending=False) == 2].reset_index(drop=True)
)

nth_lowest = (
    df[df.groupby("Dish Name")["Typical Price (USD)"]
       .rank(method="dense", ascending=True) == 3].reset_index(drop=True)
)

display(nth_highest)
display(nth_lowest)
```

### Find top three costliest dishes per country

```python
highest_per_dish = df.loc[
    df.groupby(["Country", "Dish Name"])["Typical Price (USD)"].idxmax()
]

highest_per_dish["Price_rank"]= (highest_per_dish.groupby("Country")["Typical Price (USD)"]
                                 .rank(method="dense", ascending=False))

top_three_dishes = (
    highest_per_dish[
        highest_per_dish["Price_rank"].isin([1, 2, 3])
    ]
    .sort_values(["Country", "Price_rank"])
    .reset_index(drop=True)
)

display(top_three_dishes)
                
```

### Calculate each dish’s percentage difference from the country average

```python
new_df = df.copy()

new_df["country_average"] = (
    new_df.groupby("Country")["Typical Price (USD)"]
          .transform("mean").reset_index(drop=True)
)

new_df["dish_avg"] = (
    new_df.groupby("Dish Name")["Typical Price (USD)"]
          .transform("mean").reset_index(drop=True)
)

new_df["diff_from_avg"] = (
    (
        abs(new_df["dish_avg"] - new_df["country_average"])
    )
    / new_df["country_average"]
    * 100
).round(2)

display(new_df[["Dish Name", "diff_from_avg"]])
```

### Find countries with above-average prices and below-average dish counts

```python
new_df= df.copy()
new_df["average_price"]= new_df.groupby("Country" )["Typical Price (USD)"].transform("mean")
new_df["dish_count"]= new_df.groupby("Country")["Dish Name"].transform("nunique")

result= new_df[
    (new_df["average_price"]>new_df["average_price"].mean()) &
    (new_df["dish_count"]<new_df["dish_count"].mean())
]

result= result[["Country","average_price","dish_count"]].drop_duplicates().reset_index(drop=True)
print(f"Average price: {new_df['average_price'].mean()}")
print(f"Average Dish Count: {new_df['dish_count'].mean()}")
display(result)
```

### Create a summary table containing country, dish count, average price, minimum price, and maximum price

```python
country_ingredients = {}

for country, group in df.groupby("Country"):
    ingredients = []

    for ingredient_text in group["Ingredients"].dropna():
        ingredients.extend(
            ingredient.strip()
            for ingredient in ingredient_text.split(",")
        )

    country_ingredients[country] = ingredients

print(country_ingredients)
```

### Find the most common ingredient for each country.


```python
country_ingredients = {}

for country, group in df.groupby("Country"):
    ingredient_counts = {}

    for ingredient_text in group["Ingredients"].dropna():
        for ingredient in ingredient_text.split(","):
            ingredient = ingredient.strip()

            if ingredient in ingredient_counts:
                ingredient_counts[ingredient] += 1
            else:
                ingredient_counts[ingredient] = 1

    country_ingredients[country] = ingredient_counts


most_common = {}

for country, ingredient_counts in country_ingredients.items():
    common_ingredient = None
    highest_count = 0

    for ingredient, count in ingredient_counts.items():
        if count > highest_count:
            highest_count = count
            common_ingredient = ingredient

    most_common[country] = {
        "ingredient": common_ingredient,
        "count": highest_count
    }

most_common_df = pd.DataFrame.from_dict(
    most_common,
    orient="index"
).reset_index()

most_common_df = most_common_df.rename(columns={"index": "Country"})

display(most_common_df)

```

```python
ingredients_df = (
    df[["Country", "Ingredients"]]
    .dropna()
    .assign(Ingredient=lambda x: x["Ingredients"].str.split(","))
    .explode("Ingredient")
)

ingredients_df["Ingredient"] = ingredients_df["Ingredient"].str.strip()

ingredient_counts = (
    ingredients_df
    .groupby(["Country", "Ingredient"])
    .size()
    .reset_index(name="count")
)

most_common_df = (
    ingredient_counts.loc[
        ingredient_counts.groupby("Country")["count"].idxmax()
    ]
    .sort_values("Country")
    .reset_index(drop=True)
)

display(most_common_df)
```
